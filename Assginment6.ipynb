### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data["Gross"] = pd.to_numeric(data["Gross"].astype(str).str.replace(",", ""), errors="coerce")

drama = data[data["Genre"].str.contains("Drama")].copy()
drama["Type"] = "Drama"

action = data[data["Genre"].str.contains("Action")].copy()
action["Type"] = "Action"

df = pd.concat([drama, action])

sns.boxplot(x="Type", y="Gross", data=df)
plt.show()
```
